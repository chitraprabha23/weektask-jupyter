```python
df.dropna(inplace=True)
```

```python
numeric_df=df.select_dtypes(include="number")

result=pd.DataFrame({
    "dd":numeric_df.mean(),
    "cd":numeric_df.median(),
    "cp":numeric_df.mode().iloc[0]
})
print(result.round(3))
```

```python
numeric_df=df.select_dtypes(include="number")

result=pd.DataFrame({
    "mean":numeric_df.mean(),
    "median":numeric_df.median(),
    "mode":numeric_df.mode().iloc[0]
})
print(result.round())
```

```python
result1=pd.DataFrame({
    "Variance":numeric_df.var(),
    "Standard Deviation":numeric_df.std()
})
print(result1)
```

```python
result2=pd.DataFrame({
    "Minimum":numeric_df.min(),
    "Maximum":numeric_df.max(),
    "Range":numeric_df.max()-numeric_df.min()
})
print(result2.astype(str))
```

```python
result3=pd.DataFrame({
    "Q1 (25%)":numeric_df.quantile(0.25),
    "Q2 (50%)":numeric_df.quantile(0.50),
    "Q3 (75%)":numeric_df.quantile(0.75)
})
result3["IQR"]=result3["Q3 (75%)"]-result3["Q1 (25%)"]
print(result3)
```

```python
result4=pd.DataFrame({
    "skewness":numeric_df.skew()
})

print(result4.round(3))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns 
column = numeric_df.columns[0]


sns.histplot(numeric_df[column],kde=True)

plt.title(f"skewness distribution of {column}")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()

print("skewness:",numeric_df[column].skew())
```
